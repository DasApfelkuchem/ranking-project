```python
import numpy as np
from collections import defaultdict

Unicon21 = np.loadtxt("Unicon21.txt",delimiter=";",dtype =str)
GUC25 = np.loadtxt("GUC25.txt",delimiter=";",dtype =str)
GUC26 = np.loadtxt("GUC26.txt",delimiter=";",dtype =str)
EUCW25 = np.loadtxt("EUCW25.txt",delimiter=";",dtype =str)
EUCW26 = np.loadtxt("EUCW26.txt",delimiter=";",dtype =str)
EUCS25 = np.loadtxt("EUCS25.txt",delimiter=";",dtype =str)
NAUCC25 = np.loadtxt("NAUCC25.txt",delimiter=";",dtype =str)
BUC25 = np.loadtxt("BUC25.txt",delimiter=";",dtype =str)
BUC24 = np.loadtxt("BUC24.txt",delimiter=";",dtype =str)



EVENT_WEIGHTS = {
    "Unicon21": 5,    # World level
    "EUCW26": 2,    # European Championship
    "EUCW25": 2,    # European Championship
    "EUCS25": 2,    # National Championship
    "NAUCC25": 2,
    "GUC25": 1.0,      # National Championship
    "BUC24": 1.0,
    "BUC25": 1.0,
    "GUC26": 1.0      # National Championship
}

competitions = {
    "Unicon21": Unicon21,
    "GUC25": GUC25,
    "BUC25": BUC25,
    "BUC24": BUC24,
    "GUC26": GUC26,
    "EUCW26": EUCW26,
    "EUCW25": EUCW25,
    "EUCS25": EUCS25,
    "NAUCC25": NAUCC25
}

riders = defaultdict(lambda: {
    "event_scores": [],
    "points": 0,
    "events": 0,
    "wins": 0,
    "podiums": 0
})

for event_name, results in competitions.items():
    weight = EVENT_WEIGHTS[event_name]

    for place, rider_row in enumerate(results, start=1):
        # Extract rider name
        rider = rider_row[0] if len(rider_row.shape) > 0 else str(rider_row)
        rider = " ".join(rider.split())

        base_points = 100 / place
        event_score = base_points * weight

        riders[rider]["event_scores"].append(event_score)
        riders[rider]["events"] += 1

        if place == 1:
            riders[rider]["wins"] += 1

        if place <= 3:
            riders[rider]["podiums"] += 1
BEST_RESULT_MULTIPLIERS = [1,0.67,0.33]
# Apply best-result multipliers
for rider, stats in riders.items():
    sorted_scores = sorted(stats["event_scores"], reverse=True)
    stats["points"] = sum(
        score * BEST_RESULT_MULTIPLIERS[i]
        for i, score in enumerate(sorted_scores[:len(BEST_RESULT_MULTIPLIERS)])
    )

# -----------------------------
# Sorted ranking
# -----------------------------

ranking = sorted(
    riders.items(),
    key=lambda x: (
        -x[1]["points"],
        -x[1]["wins"],
        -x[1]["podiums"],
        x[0]
    )
)

print("=== Flatland World Ranking ===\n")

for rank, (rider, stats) in enumerate(ranking, start=1):
    print(
        f"{rank:2d}. "
        f"{rider:20s} "
        f"{stats['points']:6.2f} pts "
        f"({stats['events']} events, "
        f"{stats['wins']} wins, "
        f"{stats['podiums']} podiums)"
    )
```
